```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.svm import SVC
```

```python
# Training sentences
texts = [
    "Pratik is learning artificial intelligence",
    "Pratik is studying machine learning",
    "Pratik is learning Python programming",
    "Pratik is studying deep learning",
    "Pratik is preparing for GATE",
    "Pratik is practicing data science"
]

# Categories
labels = [
    "AI",
    "AI",
    "Programming",
    "AI",
    "Exam",
    "Data Science"
]
```

```python
vectorizer = CountVectorizer()

X = vectorizer.fit_transform(texts)

print("Vocabulary:")
print(vectorizer.get_feature_names_out())
```

```python
model = SVC(kernel='linear')

model.fit(X, labels)

print("SVM model trained successfully!")
```

```python
text = input("Enter a sentence : ")

text_vector = vectorizer.transform([text])

prediction = model.predict(text_vector)

print("Input Sentence:", text)
print("Predicted Category:", prediction[0])
```
