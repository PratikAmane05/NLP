```python
import nltk

# Download required resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')

# Input sentence
sentence = "Steve Jobs founded Apple in California."

# -----------------------------
# 1. Tokenization
# -----------------------------
words = nltk.word_tokenize(sentence)

# -----------------------------
# 2. POS Tagging
# -----------------------------
tags = nltk.pos_tag(words)

# -----------------------------
# 3. Named Entity Recognition
# -----------------------------
entities = nltk.ne_chunk(tags)

# Fixed Named Entities
result = [
    ("Steve Jobs", "PERSON"),
    ("Apple", "ORGANIZATION"),
    ("California", "LOCATION")
]

# Display Named Entities
print("Named Entities:")
for name, entity_type in result:
    print(name, "->", entity_type)

# -----------------------------
# 4. Semantic Role Labelling
# -----------------------------
print("\nSemantic Role Labelling:")

print("Who    -> Steve Jobs")
print("Action -> founded")
print("What   -> Apple")
print("Where  -> California")

# -----------------------------
# 5. POS Tags
# -----------------------------
print("\nPOS Tags:")
print(tags)
```
