```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.tree import DecisionTreeClassifier
```

```python
texts = [
    "Pratik is learning Python programming",
    "Pratik is coding in Java",
    "Pratik is studying artificial intelligence",
    "Pratik is learning machine learning",
    "Pratik is preparing for GATE exam",
    "Pratik is practicing aptitude questions"
]

labels = [
    "Programming",
    "Programming",
    "AI",
    "AI",
    "Exam",
    "Exam"
]
```

```python
vectorizer = CountVectorizer()

X = vectorizer.fit_transform(texts)

print("Vocabulary:")
print(vectorizer.get_feature_names_out())
```

```python
model = DecisionTreeClassifier(random_state=42)

model.fit(X, labels)

print("Decision Tree trained successfully!")
```

```python
text = input("Enter a sentence : ")

text_vector = vectorizer.transform([text])

prediction = model.predict(text_vector)

print("Input Sentence:", text)
print("Predicted Category:", prediction[0])
```
