```python
import nltk
from nltk.classify import NaiveBayesClassifier
from nltk.classify.util import accuracy
```

```python
# Training sentences with categories
training_data = [
    ("Pratik is happy", "positive"),
    ("Pratik is a good student", "positive"),
    ("Pratik enjoys learning AI", "positive"),
    ("Pratik is sad", "negative"),
    ("Pratik feels bad", "negative"),
    ("Pratik dislikes difficult exams", "negative")
]
```

```python
def extract_features(sentence):
    words = sentence.lower().split()
    return {word: True for word in words}

training_features = [
    (extract_features(sentence), label)
    for sentence, label in training_data
]

print(training_features)
```

```python
classifier = NaiveBayesClassifier.train(training_features)

print("Classifier trained successfully!")
```

```python
text = input("Enter a sentence : ")

features = extract_features(text)

result = classifier.classify(features)

print("Input Sentence:", text)
print("Predicted Sentiment:", result)
```
